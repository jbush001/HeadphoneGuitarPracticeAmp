How hot is our LDO going to get? 

$\Delta T = (V_{in} - V_{out})I_{out} \theta_{JA}$

$\theta_{JA} = 184 \degree C/W$ (from AP2112 datasheet page 3 <https://www.diodes.com/datasheet/download/AP2112.pdf>)

Worst case load:

STM32H7B0xB = 142 mA @25C, All peripherals enabed. Table 33: <https://www.st.com/resource/en/datasheet/stm32h7b0rb.pdf> \
ES8388 = 159 mA  tab 8.5 D/P/AVDD=3.3v, Playback/record enabled: <http://www.everest-semi.com/pdf/ES8388%20DS.pdf>


```python
v_batt_max = 4.2       # Full Charge
v_batt_min = 3.5       # Cutoff 
v_out = 3.3            # System voltage
i_load_ma = 142 + 159

# e.g. https://www.digikey.com/en/products/detail/adafruit-industries-llc/4237/10244663
battery_capacity_mah = 350

theta_ja = 184 # c/w, from AP2112 datasheet page 3 https://www.diodes.com/datasheet/download/AP2112.pdf
print(f'Max temp increase {(v_batt_max - v_out) * i_load_ma / 1000 * theta_ja}C')
print(f'Min temp increase {(v_batt_min - v_out) * i_load_ma / 1000 * theta_ja}C')

ldo_run_time = battery_capacity_mah / i_load_ma
print(f'battery life {int(ldo_run_time)}:{int(ldo_run_time * 60) % 60:02}')
```

When the system is charging, the LDO runs from VBUS, which is 5v.

```python
print(f'Charge temp increase {(5.0 - v_out) * i_load_ma / 1000 * theta_ja}C')
```

Now calculate for a buck power converter <https://www.ti.com/lit/ds/symlink/tlv62568a.pdf?ts=1767203154973>

Output Efficiency $\eta$ = 96% (@300mA Figure 12 from above)

$P_{out} = V_{out}I_{out}$ \
$P_{in} = \frac{P_{out}}{\eta}$ \
$I_{batt} = \frac{P_{in}}{V_{batt}}$


```python
battery_nominal_voltage = 3.7
capacity_wh = battery_capacity_mah / 1000 * battery_nominal_voltage
usable_energy_wh = 0.96 * capacity_wh
load_power = v_out * i_load_ma / 1000
buck_run_time = capacity_wh / load_power

print(f'battery life {int(buck_run_time)}:{int(buck_run_time * 60) % 60:02}')
```
